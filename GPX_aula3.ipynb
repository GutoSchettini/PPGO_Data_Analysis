```python
import pickle
import matplotlib.pyplot as plt
import numpy as np
```

```python
# doc = '''
# Dados de um arquivo GPX processados [tempo, latitude, longitude]
# eu, hoje
# '''
# junta = [doc, tempo, latitude, longitude]

with open('dados_GPX_reduzidos.pkl','rb') as io:
    dados = pickle.load(io)
```

```python
print(dados[0])
```

```python
tempo = dados[1]
latitude = dados[2]
longitude = dados[3]
```

```python
p1_x = longitude[0]
p1_y = latitude[0]

p2_x = longitude[10]
p2_y = latitude[10]

dx = (p1_x - p2_x)**2
dy = (p1_y - p2_y)**2
distancia = (dx + dy)**(1/2) *111120
print(distancia)
```

```python
x = np.array([1,2,4,10])
np.diff(x)
```

```python
diff_lat = np.diff(latitude)
diff_lon = np.diff(longitude)

distancia = (diff_lat**2 + diff_lon**2)**.5 * 111120

distancia_total = np.sum(distancia)
print(distancia_total)
```

```python
plt.figure(figsize=(3,3))
plt.plot(longitude, latitude, '.')
plt.plot(p1_x, p1_y, 'r.')
plt.plot(p2_x, p2_y, 'g.')
```

```python
nova_lista = []
numero = []
variavel_1 = []
variavel_2 = []
for indice, elemento in enumerate(linhas[6:]):
    quebra_linha = elemento.strip().split(',')
    #print(quebra_linha)
    nova_lista.append(quebra_linha)
    numero.append(int(quebra_linha[0]))
    variavel_1.append(float(quebra_linha[1]))
    variavel_2.append(float(quebra_linha[2]))

```

# Cálculo da média


```python
# variavel_1 = [2, 4, 6]

somatorio = 0
numero_de_elementos = 0
for elemento in variavel_1:
    somatorio = somatorio + elemento
    numero_de_elementos += 1

media = somatorio / numero_de_elementos
print(media)
```

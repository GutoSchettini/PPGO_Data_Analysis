```python
import numpy as np
import datetime
import matplotlib.pyplot as plt
import pickle
```

```python
nome = 'indices_elnino.txt'

with open(nome) as io:
    linhas = io.readlines()

idx_elnino = []
tempo = []

for li in linhas:
    liq = li.replace('\n','').split('\t')
    if len(liq) > 1:
        pega_ano = int(liq[0])
        
        indices_ano = [float(x) for x in liq[1:]]
        meses_ano = np.arange(1, 13)

        tempo_dt_ano = []
        for mes in meses_ano:
            tempo_dt_ano.append(datetime.datetime(pega_ano, mes, 15))
    
        idx_elnino = idx_elnino + indices_ano
        tempo = tempo + tempo_dt_ano

tempo = tempo[:-4]
```

```python
print(len(idx_elnino))
print(len(tempo))
```

```python
plt.figure(figsize=(10,2))
plt.plot(tempo, idx_elnino)
```

```python
junta = [tempo, idx_elnino]
with open('indices_elnino.pkl', 'wb') as io:
    pickle.dump(junta, io)
```
